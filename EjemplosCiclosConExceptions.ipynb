```python
try:
  a=5
  while (a > 0):
    print(a)
    a = int(input('Ingrese un valor: '))
except:
  print('Error')
```

```python
a=5
while (a > 0):
  print(a)
  try:
    a = int(input('Ingrese un valor: '))
  except:
    print('Error: Se esperaba un número!!!')
print('Salí del ciclo.')
```

```python
a=5
while (a > 0):
  print(a)
  try:
    a = int(input('Ingrese un valor: '))
  except ValueError as e:
    print(f'Error: {e}')
print('Salí del ciclo.')
```

```python
a=5
while (a > 0):
  print(a)
  try:
    a = int(input('Ingrese un valor: '))
  except ValueError as e:
    print(f'Error: {e}')
  else:
    print('No hice error.')
  finally:
    print('Me ejecuto CON y SIN error.')
print('Salí del ciclo.')
```
